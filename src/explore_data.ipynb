```python
import pandas as pd
import requests

print("Pandas version:", pd.__version__)
print("Requests version:", requests.__version__)
```

```python
url = "https://dummyjson.com/products?limit=0"

response = requests.get(url)

print("Status Code:", response.status_code)
```

```python
data = response.json()

print("Total products:", data["total"])
print("Products fetched:", len(data["products"]))
```

```python
len(data["products"])
```

```python
products = data["products"]
df = pd.DataFrame(products)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.head()
```

```python
df.dtypes
```

```python
df.isnull().sum()
```

```python
print("===== DATA QUALITY CHECKS =====")

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Product IDs:")
print(df["id"].duplicated().sum())

print("\nInvalid Price (<= 0):")
print((df["price"] <= 0).sum())

print("\nInvalid Rating (< 0 or > 5):")
print(((df["rating"] < 0) | (df["rating"] > 5)).sum())

print("\nNegative Stock:")
print((df["stock"] < 0).sum())

print("\nNegative Discount:")
print((df["discountPercentage"] < 0).sum())

print("\nInvalid Minimum Order Quantity (<= 0):")
print((df["minimumOrderQuantity"] <= 0).sum())
```

```python
print("===== NESTED / ARRAY COLUMNS =====")

print("\nData Types:")
print(df[["tags", "dimensions", "reviews", "meta", "images"]].dtypes)

print("\nExample Values:")
print("\nTags:", df["tags"].iloc[0])
print("\nDimensions:", df["dimensions"].iloc[0])
print("\nReviews:", df["reviews"].iloc[0])
print("\nMeta:", df["meta"].iloc[0])
print("\nImages:", df["images"].iloc[0])
```

```python
print("===== CATEGORY SUMMARY =====")
print("Number of categories:", df["category"].nunique())
print("Categories:", df["category"].unique().tolist())

print("\n===== AVAILABILITY SUMMARY =====")
print("Unique values:", df["availabilityStatus"].nunique())
print(df["availabilityStatus"].value_counts().to_dict())

print("\n===== RETURN POLICY SUMMARY =====")
print("Unique values:", df["returnPolicy"].nunique())
print(df["returnPolicy"].value_counts().to_dict())
```

```python
print("===== META DATE CHECK =====")

print("Created At examples:")
print(df["meta"].apply(lambda x: x["createdAt"]).head())

print("\nUpdated At examples:")
print(df["meta"].apply(lambda x: x["updatedAt"]).head())
```

```python
created_at = pd.to_datetime(
    df["meta"].apply(lambda x: x["createdAt"]),
    utc=True
)

updated_at = pd.to_datetime(
    df["meta"].apply(lambda x: x["updatedAt"]),
    utc=True
)

print("Invalid timestamps (updatedAt < createdAt):",
      (updated_at < created_at).sum())
```

```python
print(
    "Out of Stock products with stock > 0:",
    ((df["availabilityStatus"] == "Out of Stock") & (df["stock"] > 0)).sum()
)
```

```python
print(
    "In Stock products with stock <= 0:",
    ((df["availabilityStatus"] == "In Stock") & (df["stock"] <= 0)).sum()
)
```

```python
print("Negative weight:", (df["weight"] < 0).sum())
```

```python
print("===== MINIMUM ORDER QUANTITY =====")
print(df["minimumOrderQuantity"].describe())
```

```python
print("===== REVIEWS PER PRODUCT =====")
print(df["reviews"].apply(len).describe())
```

```python
review_ratings = [
    review["rating"]
    for reviews in df["reviews"]
    for review in reviews
]

print("Total reviews:", len(review_ratings))
print("Invalid review ratings:", 
      sum((r < 0 or r > 5) for r in review_ratings))
```

```python
review_fields = ["rating", "comment", "date", "reviewerName", "reviewerEmail"]

print("===== REVIEW FIELD COMPLETENESS =====")

for field in review_fields:
    missing = sum(
        field not in review or review[field] is None
        for reviews in df["reviews"]
        for review in reviews
    )
    print(f"{field}: {missing} missing")
```

```python
print("===== NESTED FIELD COMPLETENESS =====")

for field in ["dimensions", "meta"]:
    print(f"\n{field}:")

    keys = set()
    for item in df[field]:
        keys.update(item.keys())

    for key in sorted(keys):
        missing = sum(
            key not in item or item[key] is None
            for item in df[field]
        )
        print(f"  {key}: {missing} missing")
```

```python
print("===== ARRAY FIELD CHECK =====")

print("Products with no tags:",
      df["tags"].apply(len).eq(0).sum())

print("Products with no images:",
      df["images"].apply(len).eq(0).sum())
```

```python
print("===== BRAND ANALYSIS =====")

print("Missing brand:", df["brand"].isnull().sum())
print("Present brand:", df["brand"].notnull().sum())
print("Unique brands:", df["brand"].nunique())

print("\nTop brands:")
print(df["brand"].value_counts().head(15))
```
